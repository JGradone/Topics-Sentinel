## New code!!

I prepared these data files to have a few extra columns. They now have columns [dive, climb, and profile_num]. If the glider is diving, the 'dive' column = 1. If the glider is climbing, the 'dive' column = 0 and the 'climb' column = 1. Each dive and climb is considered a profile. I have kept a running tally of the number of profiles in the 'profile_num' column.

You can subset to just a dive, climb, or specific profile by doing the following:



```

```
